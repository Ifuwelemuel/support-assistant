```python
class TicketError(Exception):
    pass

try:
    try:
        int("abc")
    except ValueError:
        raise TicketError("bad row")          # case 1
except TicketError as e:
    print("cause:", e.__cause__)
    print("context:", e.__context__)
```

```python
class TicketError(Exception):
    pass

try:
    try:
        int("abc")
    except ValueError as exc:
        raise TicketError("bad row") from exc          # case 1
except TicketError as e:
    print("cause:", e.__cause__)
    print("context:", e.__context__)
```

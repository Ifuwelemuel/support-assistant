# pandas vs plain python

```python
from pathlib import Path

from support_assistant.analysis import load_tickets, count_by,high_priority,top_words
from support_assistant.data import load_tickets as load_df



#ROOT = Path.cwd().parents[2]
#ROOT
ROOT = Path.cwd().parents[2] if Path.cwd().name == "notebooks" else Path.cwd()
SAMPLE = ROOT / "data" / "sample" / "ticket.csv"

```

```python
print(SAMPLE)
SAMPLE.exists()
#/Users/mac/Desktop/2026/Bootcamp/code/support-assistant/data/sample/ticket.csv
```

```python
tickets = load_tickets(SAMPLE)
df = load_df(SAMPLE)
print(len(tickets), "tickets as objects;", df.shape, "as a table")
df.head(3)
```

```python
for field in ["category","priority","channel"]:
    plain =count_by(tickets,field)
    pandas_ = df[field].value_counts().to_dict()
    assert plain == pandas_, field
    print(f"{field:9} {pandas_}")
print("all counts match")
```

```python
plain_ids = [t.id for t in high_priority(tickets)]
pandas_ids = df[df["priority"] == "high"] ["id"].tolist()

assert plain_ids == pandas_ids
print("high_priority:", pandas_ids)

df.groupby(["channel", "priority"]).size()
```

```python
words = df["subject"].str.lower().str.split().explode()
pandas_top = words.value_counts().head(2).to_dict()
plain_top = dict(top_words(tickets, 2))
assert pandas_top == plain_top

print("top word:", pandas_top)
```

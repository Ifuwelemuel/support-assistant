```python
from dataclasses import dataclass
@dataclass
class Ticket:
    id: str
    created_at:str
    channel:str
    category: str
    priority: str
    subject: str
    body: str


    def is_high(self):
        return self.priority == "high"


    def describe(self):
        return f"{self.id} ({self.priority}) {self.subject}"


```

```python
rows = load_rows()
t = Ticket(**rows[0])
t
```

```python
counts = {}
for row in rows:
    for word in row["subject"].lower().split():
        counts[word] = counts.get(word,0) + 1

len(counts)
```

```python
counts["export"]
```

```python
counts["change"]
```

```python
counts.items()
```

```python
def count_of(pair):
    return pair[1]

ranked = sorted(counts.items(), key=count_of,reverse=True)
ranked[:3]
```
